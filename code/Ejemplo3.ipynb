```julia
using JuMP
using HiGHS
using Ipopt
using Optimization
using Optim
```

```julia
# Set LP problem
model = Model()

# variables 
@variable(model, x1 >= 0)
@variable(model, x2 >= 0)
@variable(model, x3 >= 0)
@variable(model, x4 >= 0)

# constraints 
@constraint(model,  x1 + 2x2 + 2x3 + 4x4 <= 40)
@constraint(model, 2x1 -  x2 +  x3 + 2x4 <= 8)
@constraint(model, 4x1 - 2x2 +  x3 -  x4 <= 10)

# objective function
@objective(model, Max, 2x1 + x2 - x3 + 5x4)
```

```julia
set_optimizer(model, HiGHS.Optimizer)
```

```julia
model
```

```julia
optimize!(model)
```

```julia
print("x1 = ", value.(x1), "\n")
print("x2 = ", value.(x2), "\n")
print("x3 = ", value.(x3), "\n")
print("x4 = ", value.(x4), "\n\n")

print("z = ", objective_value(model))
```
